2\. Find the outer product of the following two vecotrs

```python
u = np.array([1,3,5,7])
v = np.array([2,4,6,8])
```

Do this in the following ways:

   * Using the function outer in numpy
   * Using a nested for loop or list comprehension
   * Using numpy broadcasting operatoins


````python
u = np.array([1,3,5,7])
v = np.array([2,4,6,8])

w = np.outer(u,v)
print(w)

w = np.array([i*j for i in u for j in v]).reshape(4,4)
print(w)

a = np.tile(u,(4, 1)).T
b = a*v
print(b)
````

````python
u = np.array([1,3,5,7])
v = np.array([2,4,6,8])
print(np.outer(u,v))

l = [i * j for i in u for j in v]
m = np.array(l).reshape(4,4)
print(m)

u_1 = np.tile(u, (4,1)).T
outer= u_1*v
print(outer)
````

````python
import numpy as np

u = np.array([1,3,5,7])
v = np.array([2,4,6,8])

print("The outer product of the 2 arrays with the outer function in numpy is: \n", np.outer(u,v))

p = np.ones((len(u), len(v)))
for i in range(len(u)):
    for j in range(len(v)):
        p[i,j] = u[i]*v[j]

print("The outer product of the 2 arrays with the nested for loop is: \n", p)

b = np.tile(u, (len(v),1)).T
print("The outer product of the 2 arrays with the broadcasting operations is: \n", b*v)
````

````python
import time
import matplotlib.pyplot as plt

def find_prime(N):
    numbers = np.arange(0, N+1, 1)
    mask = np.ones(N+1, 'bool')
    for i in range(2, N+1):
        if mask[i]==True:
            for j in range(i+1, N+1):
                if j%i == 0:
                    mask[j] = False
                else:
                    pass
        else: 
            pass
    return numbers[mask]

def sieve_erat(N):
    numbers = np.arange(0, N+1, 1)
    mask = np.ones(N+1, 'bool')
    for i in range(2, N+1):
        if mask[i]==True:
            rang = np.arange(i**2, N+1, i)
            for j in rang:
                if j%i == 0:
                    mask[j] = False
                else:
                    pass
        else: 
            pass
    return numbers[mask]

def find_time(N):
    
    start_1 = time.time()
    func = find_prime(N)
    elapse_1 = time.time()-start_1
    
    start_2 = time.time()
    func = sieve_erat(N)
    elapse_2 = time.time()-start_2
    
    return elapse_1, elapse_2

print(find_prime(100))
print(sieve_erat(100))
find_time(100)
N = np.arange(0, 10000, 1000)
time_scale_1 = []
time_scale_2 = []
for i in N:
    j, k = find_time(i)
    time_scale_1.append(j)
    time_scale_2.append(k)
    
print(time_scale_1)
print(time_scale_2)

fig, ax = plt.subplots(1,1, figsize=(6,6))
ax.plot(N, time_scale_1, 'o-', color = 'r')
ax.plot(N, time_scale_2, 'o-', color = 'b')
ax.grid(True)
plt.show()
````

````python
import time
import matplotlib.pyplot as plt

def find_prime(N):
    a = np.array([i for i in range(N+1)])
    mask = np.ones(N+1, dtype = bool)
    mask[0:2] = False
    for i in range(2, N+1):
        for j in range(i, N+1):
            k = i*j #multiples of i
            if k>N-1:
                pass
            else: 
                mask[k] = False
    return a[mask]

def find_prime_1(N):
    a = np.array([i for i in range(N+1)])
    mask = np.ones(N+1, dtype = bool)
    mask[0:2] = False
    for i in range(2, int(np.sqrt(N))+1):
        if mask[i]:
            mask[i*i:N+1:i] = False
    return a[mask]
    
def find_time(N):
    start = time.time()
    find_prime(N)
    elapse = time.time()-start
    print("Time with N= ", N, "is", elapse)
    return elapse

def find_time_1(N):
    start = time.time()
    find_prime_1(N)
    elapse = time.time()-start
    print("Time with optimization for N= ", N, "is", elapse)
    return elapse

n = [i for i in range(1, 10001, 1000) ]
time_1 = []
time_2 = []
for i in n:
    time_1.append(find_time(i))
    time_2.append(find_time_1(i))
plt.plot(n, time_1, 'r-', n, time_2, 'g-')
plt.grid(True)
plt.show()
````

8\. Diffusion using random walk

Consider a simple random walk process: at each step in time, a walker jumps right or left (+1 or -1) with equal probability. The goal is to find the typical distance from the origin of a random walker after a given amount of time. 
To do that, let's simulate many walkers and create a 2D array with each walker as a raw and the actual time evolution as columns

  * Take 1000 walkers and let them walk for 200 steps
  * Use randint to create a 2D array of size walkers x steps with values -1 or 1
  * Build the actual walking distances for each walker (i.e. another 2D array "summing on each raw")
  * Take the square of that 2D array (elementwise)
  * Compute the mean of the squared distances at each step (i.e. the mean along the columns)
  * Plot the average distances (sqrt(distance\*\*2)) as a function of time (step)
  
Did you get what you expected?

````python
step= [-1,1]
np.random.seed(123)
m = np.random.choice(step, (1000,200))
print(m)
dist = np.cumsum(m, axis=1)
print(dist)
print(dist**2)
mean = np.mean(dist**2, axis = 0)
print(mean)
t = np.arange(0,200)
av = mean**0.5
plt.plot(t, av)
````
