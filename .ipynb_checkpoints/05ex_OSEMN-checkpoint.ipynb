1\. Create a random list of number and then save it to a text file named "simple_data.txt"

```python
import numpy.random as npr

random_numbers = npr.rand(100)
print(random_numbers)
new_file = "simple_data.txt"
with open(new_file, 'w') as f:
    for number in random_numbers:
        f.write(f"{number}\n")
print("Data saved to simple_data.txt")
```

```python
import random

random_numbers = [random.randint(1, 100) for _ in range(20)]

file_name = "simple_data.txt"
with open(file_name, "w") as f:
    for number in random_numbers:
        f.write(f"{number}\n")

print("Data saved to simple_data.txt")
```

2\. Create a random matrix of 5x5 and then save it to a text file named "data.txt"

```python
m = npr.rand(5,5)
new_file = "data.txt"
with open(new_file, 'w') as f:
    for row in m:
        f.write(" ".join(f"{val:.4f}" for val in row) + "\n")
!more data.txt
```

3\. Load the saved txt file of point 2 and convert it to a csv file (by hand)

```python
csv_file = "data.csv"
data = []
with open("data.txt", 'r') as f:
    for row in f:
        line = row.split()
        data.append(line)

print(data)
with open(csv_file, 'w') as f:
    for row in range(len(data)):
        f.write(",".join(f"{data[row][val]}" for val in range(5))+ "\n")

!more data.csv
```

4\. load the binary file named *credit_card.dat* and convert the data into the real credit-card number.
Each line correspond to a credit card number.
Each character is composed by 6 bit (even the space) and the last 4 bit are just a padding

**hint**: use the `chr()` function to convert a number to a char

```python
a = [1,2,3,4]
num = 0
for i in a:
    num +=1
print(num)
print(len(a)-1)
```

```python
import numpy as np

data = np.loadtxt('credit_card.dat')
file = 'credit_card.dat'
with open(file, 'rb') as f: # 'rb' is needed to read the file with a precise length step
    for row in f:
        dig = []
        num_counter = 0
        for i in range(0, len(row)-4, 6):
            num_counter += 1
            chunk = row[i:i+6]
            value = int(chunk, 2)
            if num_counter % 5 == 0:
                dig.append(" ")
            else:
                dig.append(str(value))
        credit_num = "".join(dig)
        print(credit_num)
```
