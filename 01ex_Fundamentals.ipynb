2\. Convert the following function into a pure function with no global variables or side effects

````python
def f(alist, x):
    new_list = alist.copy()
    for i in range(x):
         new_list.append(i)
    return new_list

alist = [1,2,3]
x = 5
ans = f(alist, x)
print (ans)
print (alist) # alist has been changed!
````

3\. Write a `decorator` hello that makes every wrapped function print “Hello!”, i.e. something like:

```python
@hello
def square(x):
    return x*x
```



````python
def hello(func):
    def wrapper(*args, **kwargs):
        print("Hello!")
        func(*args, **kwargs)
        print("Hello!")
    return wrapper

@hello
def square(x):
    print(x*x)
    
square(2)
````

````python
def hello(func):
    def wrapper(*args, **kwargs):
        print("Hello!")
        func()
        print("Hello!")
    return wrapper

@hello
def square():
    print(x**2)

square(3)
````

4\. Write the factorial function so that it a) does and b) does not use recursion.

````python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)

n = 3
print(factorial(3))
````

````python
#a) with recursion
def factorial(n):
    if n==0:
        return 1
    else:
        return n*factorial(n-1)
print(factorial(4))

#b) without recursion
def factorial_2(n):
    f = 1
    if n == 0:
        return 1
    else:
        for i in range(n):
            f*=n-i
    return f
print(factorial_2(4))
````

````python
#a) with recursion
def factorial(x):
    if x == 0:
        return 1
    else:
        return x*factorial(x-1)
        
print(factorial(1))
````

````python
#b) no recursion
def factorial(x):
    if x == 0:
        return 1
    else: 
        for i in range(1,x):
            x *= i
        return x
        

print(factorial(5))
    
````

````python
#a) with recursion
def factorial(n):
    if n==0 or n==1:
        return 1
    else:
        return n*factorial(n-1)

x=int(input("Set the x value: "))
print ("The factorial of x with recursion is: ", factorial(x))
````

````python
#b) without recursion
def factorial(n):
    if n==0 or n==1:
        return 1
    else:
        fact=1
        for i in range(1, n+1):
            fact *= i
        return fact

x=int(input("Set the x value: "))
print ("The factorial of x without recursion is: ", factorial(x))
````

5\. Use HOFs (zip in particular) to compute the weight of a circle, a disk and a sphere, assuming different radii and different densities:

```python
densities = {"Al":[0.5,1,2],"Fe":[3,4,5],"Pb": [15,20,30]}
radii = [1,2,3]
```

where the entries of the dictionary's values are the linear, superficial and volumetric densities of the materials respectively.

In particular define a list of three lambda functions using a comprehension that computes the circumference, the area and the volume for a given radius.


````python
#Da rifareee
import math

densities = {"Al":[0.5,1,2],"Fe":[3,4,5],"Pb": [15,20,30]}
radii = [1,2,3]

#lambdas =
l = [(lambda x: 2*math.pi*x), (lambda x: math.pi*x**2), (lambda x: (3/4)*math.pi*x**3)]

label = ("Disk", "Circle", "Sphere")

for i, j in densities.items():
    print("Material ", i, ":")
    for labels, (rho, f) in zip(label,(zip(j,l))):
        weight = [rho*f(x) for x in radii]
        print(labels, "=", weight)
````
